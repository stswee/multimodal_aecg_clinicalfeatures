```python
import numpy as np
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
)

def calculate_metrics(data, task):
    label_col = f"{task}_label"
    probability_col = f"{task}_probability"

    applicable = data[label_col].notna()
    y = data.loc[applicable, label_col].astype(int).to_numpy()
    p = data.loc[applicable, probability_col].to_numpy()

    return {
        "roc_auc": roc_auc_score(y, p),
        "pr_auc": average_precision_score(y, p),
        "brier_score": brier_score_loss(y, p),
        "n": len(y),
        "events": int(y.sum()),
    }


def bootstrap_confidence_intervals(
    predictions,
    task,
    n_bootstrap=5000,
    seed=42,
):
    rng = np.random.default_rng(seed)
    n_patients = len(predictions)

    estimates = {
        "roc_auc": [],
        "pr_auc": [],
        "brier_score": [],
    }

    for _ in range(n_bootstrap):
        # Resample patients, not ECG segments
        sampled_indices = rng.integers(
            0,
            n_patients,
            size=n_patients,
        )
        sampled = predictions.iloc[sampled_indices]

        try:
            metrics = calculate_metrics(sampled, task)
        except ValueError:
            # Skip a replicate if an endpoint has only one class
            continue

        for metric in estimates:
            estimates[metric].append(metrics[metric])

    point = calculate_metrics(predictions, task)
    rows = []

    for metric in estimates:
        values = np.asarray(estimates[metric])

        rows.append({
            "Outcome": task.upper(),
            "Metric": metric,
            "Estimate": point[metric],
            "CI lower": np.percentile(values, 2.5),
            "CI upper": np.percentile(values, 97.5),
            "Bootstrap replicates": len(values),
            "N": point["n"],
            "Events": point["events"],
        })

    return pd.DataFrame(rows)


performance_with_ci = pd.concat(
    [
        bootstrap_confidence_intervals(
            predictions,
            task="scd",
            n_bootstrap=5000,
            seed=42,
        ),
        bootstrap_confidence_intervals(
            predictions,
            task="pfd",
            n_bootstrap=5000,
            seed=43,
        ),
    ],
    ignore_index=True,
)

display(performance_with_ci)

performance_with_ci.to_csv(
    root / "pooled_outer_test" /
    "ecg_pooled_performance_with_95ci.csv",
    index=False,
)
```
